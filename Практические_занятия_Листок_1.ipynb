## 1.1

```python
A = sp.Matrix([
    [1, 5, -2, 0],
    [-3, 1, 9, -5],
    [4, -8, -1, 7]
])

b = sp.Matrix([-7, 9, 0])
rref_matrix, main_columns = A.row_join(b).rref()

t = sp.symbols("t")

answer = (
    -sp.Rational(11, 7) - sp.Rational(8, 7)*t,
    -sp.Rational(6, 7) + sp.Rational(2, 7)*t,
    sp.Rational(4, 7) + sp.Rational(1, 7)*t,
    t
)

print("rref:")
sp.pprint(rref_matrix)
print("Ответ:")
sp.pprint(answer)
```

## 1.7

```python
n = 80000
diagonal = np.linspace(0.99, 1.01, n)

a11 = diagonal[0]
a22 = diagonal[1]

det_block = a11 * a22 - 0.57 * 0.03
log_product = np.sum(np.log(diagonal[2:]))

det_A = det_block * np.exp(log_product)
trace_inverse = (a11 + a22) / det_block + np.sum(1 / diagonal[2:])

print("Ответ:")
print("1)", round(det_A, 3))
print("2)", round(trace_inverse, 3))
```

## 1.8

```python
n = 43000

q = 1.07 ** (1 / (n - 1))
diagonal = q ** np.arange(n)
first_row = np.linspace(1, 1.10, n)

inverse_diagonal = 1 / diagonal
inverse_first_row = -first_row[1:] / diagonal[1:]

min_element = min(np.min(inverse_diagonal), np.min(inverse_first_row))
sum_elements = np.sum(inverse_diagonal) + np.sum(inverse_first_row)

print("Ответ:")
print("1)", round(min_element, 3))
print("2)", round(sum_elements, 1))
```

## 1.9

```python
gamma = 0.057

A = np.array([
    [7 * gamma, 14 * gamma],
    [9 * gamma, 1 * gamma]
])

I = np.eye(2)

S = np.zeros((2, 2))
power = np.eye(2)

for i in range(71):
    if i >= 4:
        S = S + power
    power = power @ A

answer1 = S.max()

B4 = np.linalg.inv(I - A) @ np.linalg.matrix_power(A, 4)
answer2 = B4.max()

answer3 = None

for k in range(1, 200):
    Bk = np.linalg.inv(I - A) @ np.linalg.matrix_power(A, k)
    if Bk.max() < 0.0002:
        answer3 = k
        break

print("Ответ:")
print("1)", round(answer1, 6))
print("2)", round(answer2, 6))
print("3)", answer3)
```

```python
x = sp.symbols("x")

M = sp.Matrix([
    [7, 5, 5, x],
    [2, 2, 6, 9],
    [3, x, 8, 0],
    [3, 1, 1, 0]
])

det_M = sp.expand(M.det())
poly = sp.Poly(det_M, x)

a = poly.coeff_monomial(x**2)
b = poly.coeff_monomial(x)

print("Определитель:")
sp.pprint(det_M)
print("Ответ:")
print("1)", a)
print("2)", b)
```

## 1.11

```python
n = 100
matrix = np.zeros((n, n), dtype=int)

for i in range(n):
    for j in range(i + 1):
        matrix[i, j] = comb(i, j) % 2

colors = ListedColormap(["#f2c66d", "#4f64b0"])

plt.figure(figsize=(7, 7))
plt.imshow(matrix, cmap=colors, origin="upper")
plt.title("n = 100")
plt.show()

print("Ответ: матрица построена")
```

## 1.12

```python
H1 = np.array([[1]])

H2 = np.array([
    [1, 1],
    [1, -1]
])

H4 = np.kron(H2, H2)

print("H1:")
print(H1)

print("H2:")
print(H2)

print("H4:")
print(H4)

print("H4 = H2 ⊗ H2:", np.array_equal(H4, np.kron(H2, H2)))

H64 = np.array([[1]])

for i in range(6):
    H64 = np.kron(H64, H2)

colors = ListedColormap(["#4058d6", "#ff8c1a"])

plt.figure(figsize=(7, 7))
plt.imshow(H64, cmap=colors)
plt.title("H64")
plt.show()

print("Ответ: H64 построена")
```

## 1.13

```python
A = np.array([
    [14, 10, 21],
    [12, 15, 26],
    [18, 11, 27]
], dtype=float)

values, vectors = np.linalg.eig(A)

index = np.argmax(values.imag)

Z = values[index]
X = vectors[:, index]
X = X / X[0]

print("Z =", Z)
print("X =", X)

print("Ответ:")
print("1)", round(Z.imag, 6))
print("2)", round(X[1].real, 6))
```

## 1.15

```python
A = sp.Matrix([
    [17, -8, 4],
    [-8, 17, -4],
    [4, -4, 11]
])

e1 = sp.Matrix([1/sp.sqrt(2), 1/sp.sqrt(2), 0])
e2 = sp.Matrix([-1/(3*sp.sqrt(2)), 1/(3*sp.sqrt(2)), 4/(3*sp.sqrt(2))])
e3 = sp.Matrix([sp.Rational(2, 3), sp.Rational(-2, 3), sp.Rational(1, 3)])

P = sp.Matrix.hstack(e1, e2, e3)
D = sp.simplify(P.T * A * P)

print("Базис:")
sp.pprint(P)

print("Ответ:")
sp.pprint(D)
```
